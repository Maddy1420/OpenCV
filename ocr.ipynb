```python
import cv2
import pytesseract
import imutils

pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'
```

```python
image_path = r"C:\Users\aarun\Agentic AI\DeepLearning\OpenCV\numberplate.jpg"
image = cv2.imread(image_path)

resized_image = imutils.resize(image, width=600)
cv2.imshow('Original Image', resized_image)
cv2.waitKey(0)  
cv2.destroyAllWindows()
```

```python
gray_image = cv2.cvtColor(resized_image, cv2.COLOR_BGR2GRAY)
cv2.imshow('Grayscale Image', gray_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray_image = cv2.bilateralFilter(gray_image, 11, 17, 17)
cv2.imshow('Filtered Image', gray_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
edges = cv2.Canny(gray_image, 30, 200)
cv2.imshow('Canny Edges', edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cnts,new = cv2.findContours(edges.copy(), cv2.RETR_LIST, cv2.CHAIN_APPROX_SIMPLE)
image_copy = resized_image.copy()
cv2.drawContours(image_copy, cnts, -1, (0, 255, 0), 3)
cv2.imshow('Contours', image_copy)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cnts = sorted(cnts, key=cv2.contourArea, reverse=True)[:30]
screenCnt = None
image2= resized_image.copy()
cv2.drawContours(image2, cnts, -1, (0, 255, 0), 3)
cv2.imshow('Top 30 Contours', image2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
i=7
for c in cnts:
    perimeter = cv2.arcLength(c, True)
    approx = cv2.approxPolyDP(c, 0.018 * perimeter, True)

    if len(approx) == 4:
        screenCnt = approx
        x, y, w, h = cv2.boundingRect(c)
        new_img = resized_image[y:y + h, x:x + w]
        cv2.imwrite('Cropped_Image.png', new_img)
        i += 1
        break
```

```python
cv2.drawContours(resized_image, [screenCnt], -1, (0, 255, 0), 3)
cv2.imshow('Detected Number Plate', resized_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
